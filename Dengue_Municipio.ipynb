# Análise de casos de dengue por município

## Pergunta norteadora
Como evoluíram, ano a ano, a contagem de casos e a taxa de infecção de dengue nos municípios do estado de São Paulo entre 2013 e 2023?

## Objetivo
Identificar aumento ou diminuição dos casos de dengue ao longo dos anos, para apoiar a área da saúde na tomada de providências na condução dos casos.

```python
# Bibliotecas importantes

# Manipulação dos dados
import pandas as pd
import numpy as np

# Visualização dos dados
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

### Abrindo o excel

```python
df_dengue = pd.read_excel('Base de Dados Dengue.xlsx')
df_dengue.head()
```

```python
## Filtrando na base de dados pra trazer apenas uf = SP e seus municípios

df_dengue_sp = df_dengue[df_dengue['uf'] == 'SP']
df_dengue_sp
```

## Criando coluna ano

```python
## Criando coluna de ano

df_dengue_sp['ano'] = df_dengue_sp['data_infeccoes'].dt.year
df_dengue_sp
```

## Trazendo a qtd de infecções pela UF SP

```python
infeccoes_por_municipio = df_dengue_sp.groupby(['municipio', 'uf' , 'ano'])['qtd_infeccoes'].sum().reset_index()
infeccoes_por_municipio
```

# Conclusões da exploração

## Estrutura e qualidade da base
Após filtrar o estado de SP, a base tem 85.140 linhas, com 645 municípios e 132 meses (de janeiro de 2013 a dezembro de 2023). Não há valores nulos nem linhas duplicadas (município + mês). A distribuição de `qtd_infeccoes` é muito assimétrica: a mediana é 0, 75% dos registros têm até 3 casos e o máximo é 19.584. Isso indica muitos meses sem casos, com poucos picos muito altos.

## Valor extremo: março de 2015
O maior valor da base é da cidade de São Paulo em março de 2015: 19.584 casos. Os meses vizinhos mostram subida e queda graduais (fevereiro: 4.246; abril: 7.244), o que sugere uma epidemia e não um erro de registro. Por isso, o valor foi mantido.

Segundo matéria do UOL Vestibular (Novelo Comunicação), com dados do Ministério da Saúde até 18/04/2015, 2015 teve aumento expressivo de casos de dengue no país, e o estado de SP concentrava 55% deles. A matéria indica o pico habitual entre março e abril e aponta a crise hídrica como possível fator (hipótese citada pela fonte, não verificada nesta análise).

FONTE: https://vestibular.uol.com.br/resumo-das-disciplinas/atualidades/brasil-tem-novo-surto-de-dengue-sp-concentra-a-maioria-dos-casos--brasil-tem-novo-surto-de-dengue-sp-concentra-a-maioria-dos-casos.htm

## Hipótese de sazonalidade (a verificar na etapa 3)
Na cidade de São Paulo, os picos de 2013 e 2015 ocorreram em março e abril. Nos meses observados de 2023 (agosto a dezembro), o maior valor foi em dezembro. Isso sugere que o período de pico pode variar entre os anos. A relação com chuvas e calor do verão é uma hipótese, a ser confirmada agrupando os dados mensalmente em todos os anos.

## Premissas e limitações
- Premissa: `qtd_infeccoes` representa os casos confirmados de dengue, conforme o enunciado do desafio.
- Limitação: um zero pode significar ausência de casos ou falta de registro; a base não permite distinguir.
- Limitação: a base não contém população (confirmado com a turma no Discord). O cálculo da taxa de infecção depende de uma fonte externa de população.
